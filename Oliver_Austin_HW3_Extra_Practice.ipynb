## 1. ISLR 3.7.3

Reference: [*An Introduction to Statistical Learning*](https://www.statlearning.com/s/ISLRSeventhPrinting.pdf),
Section 3.7, Exercise 3 (p. 120).

Let $G$ be GPA, $Q$ be IQ, and $D$ be the gender indicator used in the
exercise (female = 1, male = 0). The fitted salary, in thousands of
dollars, is

$$\widehat{Y}=50+20G+0.07Q+35D+0.01GQ-10GD.$$

### (a) Comparing fitted salaries

At the same GPA and IQ, the difference between the female and male
predictions is

$$\widehat{Y}_{D=1}-\widehat{Y}_{D=0}=35-10G.$$

This difference is negative when $G>3.5$, so **option iii** is correct:
the model predicts a higher salary for males when GPA is sufficiently
high. The predictions are equal at $G=3.5$; below that, the female
prediction is higher. These comparisons describe the given model.

### (b) Prediction for GPA 4.0 and IQ 110

Substituting $G=4$, $Q=110$, and $D=1$ gives

$$\widehat{Y}=50+20(4)+0.07(110)+35+0.01(4)(110)-10(4)=137.1.$$

The predicted starting salary is **\$137,100**.

### (c) Interpreting the interaction coefficient

**False.** A small coefficient does not imply weak evidence for an
interaction. Its size depends on the units of the predictors. For
example, the $GQ$ term contributes $0.01(4)(110)=4.4$ thousand dollars
in part (b). The GPA slope also depends on IQ:

$$\frac{\partial\widehat{Y}}{\partial G}=20+0.01Q-10D.$$

To judge statistical significance, I would need the coefficient's
standard error or a confidence interval, neither of which is given.

## 2. ISLR 3.7.5

Reference: [*An Introduction to Statistical Learning*](https://www.statlearning.com/s/ISLRSeventhPrinting.pdf),
Section 3.7, Exercise 5 (p. 121).

For least squares regression without an intercept, the estimated slope is

$$\widehat{\beta}=\frac{\sum_{j=1}^{n}x_jy_j}{\sum_{k=1}^{n}x_k^2},$$

assuming $\sum_k x_k^2>0$. Substituting into the fitted value gives

$$\widehat{y}_i=x_i\widehat{\beta}
=\frac{x_i\sum_{j=1}^{n}x_jy_j}{\sum_{k=1}^{n}x_k^2}
=\sum_{j=1}^{n}\left(\frac{x_ix_j}{\sum_{k=1}^{n}x_k^2}\right)y_j.$$

Therefore, for a fixed observation $i$,

$$\boxed{a_j=\frac{x_ix_j}{\sum_{k=1}^{n}x_k^2}}.$$

The weights depend only on the predictor values, so $\widehat{y}_i$ is
a linear combination of the observed responses. They do not have to be
positive or sum to one. If all predictor values are zero, the slope
formula is undefined, although all fitted values are zero.
